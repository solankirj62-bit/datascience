<h2>Chacking And Handaling Missing Values

```python
# checking missing values
print(df.isnull().sum())
sns.heatmap(df.isnull(),yticklabels=False,cbar=False)
```

```python
# fill missing vlues by most repeting value in column (columns datatype is object)

df['rating'] = df['rating'].fillna(df['rating'].mode()[0])
# df['rating'].fillna(df['rating'].mode()[0], inplace=True)
df['duration'] = df['duration'].fillna(df['duration'].mode()[0])
df['country'] = df['country'].fillna(df['country'].mode()[0])
```

```python
# There are large amount of missing values in director column we cant delete that,so we put Director in place 
df['director'] = df['director'].fillna('Director')
df['cast'] = df['cast'].fillna('Hero Heroine')
```

```python
# delete raw which has missing value in date_added coluns(there are few raws (10))
df.dropna(subset='date_added',inplace=True)
```
